# 01 — Problem Statement

## Project Title

Automated Translation of Organizational Policies into Computational Rules Using Artificial Intelligence

## Problem Statement

Organizational policies are written in natural language and may contain complex conditions, dependencies, and exceptions. Translating these policies into computational rules using artificial intelligence can introduce logical errors or omit requirements, even when the generated rules execute successfully (Gupta & Sreenivasamurthy, 2026).

## Motivation

Organizations require reliable computational rules to automate operational processes and enforce organizational requirements. Errors introduced during AI-based policy translation can result in incorrect decisions, inconsistent enforcement, and unintended outcomes.

This research focuses on introducing an independent constraint-validation mechanism to evaluate whether AI-generated computational rules preserve the requirements expressed in the original organizational policies.

## Project Objective

Evaluate whether independent automated constraint validation improves the detection of logical errors in AI-generated computational rules.

## Reference

Gupta, V., & Sreenivasamurthy, D. (2026). *Prose2Policy (P2P): A practical LLM pipeline for translating natural-language access policies into executable Rego.* https://arxiv.org/abs/2603.15799
