# 02 — Background

## Organizational Context
Organizations express operational and access-control requirements in natural language. Automated enforcement requires precise computational rules, but ambiguity complicates formalization (Cosler et al., 2023).

## Existing Research
Prose2Policy translates natural-language access-control policies into executable Rego. The published study reports 95.3% compilation success, 82.2% positive-test success, and 98.9% negative-test success (Gupta & Sreenivasamurthy, 2026). These are published baseline results, not capstone experimental results.

## Specific Problem
A rule may compile successfully while omitting conditions or producing unintended access decisions. Compilation and limited behavioral testing do not establish full semantic correctness.

## Technical Foundation
- Python: data processing and implementation.
- Large language models: policy translation.
- Rego: computational policy representation.
- Open Policy Agent (OPA): execution and behavioral testing.
- Microsoft Z3: formal constraint validation.
- Google Colab and GitHub: development and documentation.

## Research Focus
Evaluate whether independent Z3 constraint checking identifies semantic errors missed by baseline validation while keeping generated Rego rules unchanged.

## Reference
Gupta, V., & Sreenivasamurthy, D. (2026). *Prose2Policy (P2P): A practical LLM pipeline for translating natural-language access policies into executable Rego.* https://arxiv.org/abs/2603.15799
