# 04 — Related Work

## 4.1 Prose2Policy
Gupta and Sreenivasamurthy (2026) provide the baseline system architecture for translating natural-language access-control policies into executable Rego rules. The methodology includes policy preprocessing, component extraction, schema validation, and Rego generation and testing.

## 4.2 Logic.py
Kesseli, O’Hearn, and Cabral (2025) examine the relationship between language models and constraint solvers. This work supports the proposed integration of formal constraint validation using Microsoft Z3.

## 4.3 Automated Requirement Contradiction Detection
Gärtner and Göhlich (2024) examine the combination of formal logic and large language models to identify contradictions in natural-language requirements. This work supports the evaluation of logical inconsistencies in generated computational rules.

## 4.4 Research Synthesis
Prose2Policy provides the baseline translation methodology, Logic.py supports formal constraint solving, and contradiction detection research supports logical error evaluation. The proposed research combines these concepts to evaluate additional semantic error detection.
