# 03 — Research

## Research Gap
Existing AI-based policy translation methods do not necessarily verify whether generated computational rules preserve the meaning and requirements of the original organizational policies (Gupta & Sreenivasamurthy, 2026).

## Research Question
To what extent can independent automated constraint validation improve the detection of logical errors and preservation of organizational policy requirements in AI-generated computational rules?

## Research Objectives
1. Examine existing AI-based policy translation methodologies.
2. Implement the Prose2Policy baseline methodology.
3. Identify limitations of compilation and behavioral testing.
4. Develop additional constraint validation using Microsoft Z3.
5. Evaluate both conditions using consistent experimental inputs.
6. Compare performance using measurable criteria.

## Research Scope
The research focuses on natural-language access-control policies from the ACRE dataset. The experimental design compares baseline validation with baseline validation supplemented by independent Z3 checking. Generated Rego rules remain unchanged between conditions.

## Reference
Gupta, V., & Sreenivasamurthy, D. (2026). *Prose2Policy (P2P): A practical LLM pipeline for translating natural-language access policies into executable Rego.* https://arxiv.org/abs/2603.15799
