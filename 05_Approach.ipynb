# 05 — Research Approach

## 5.1 Research Methodology
The proposed comparative experiment evaluates two validation conditions.

### Condition A — Prose2Policy Baseline
1. Receive natural-language policy statements.
2. Preprocess policy text.
3. Extract policy components.
4. Validate the structured schema.
5. Generate executable Rego rules.
6. Conduct compilation and behavioral testing using OPA.

### Condition B — Baseline with Additional Z3 Validation
1. Use the same policy statements and baseline-generated Rego rules.
2. Independently review original policy requirements and encode reference constraints.
3. Encode generated rule behavior as comparable formal constraints.
4. Use Z3 to check for logical differences and record counterexamples.
5. Compare additional error detection against verified reference outcomes.

## 5.2 High-Level System Architecture
```text
ACRE dataset
    -> Data collection and preparation
    -> Policy preprocessing and component extraction
    -> Schema validation and Rego generation
    -> OPA baseline validation
        |-> Condition A: baseline outputs
        |-> Condition B: baseline + Z3 validation
              ^ independently reviewed reference constraints
    -> Comparative evaluation and findings
```
